```python
from sympy import symbols, Eq, solve, pprint, I
import numpy as np
import matplotlib.pyplot as plt
# Define symbols
x, y, R, w, C, R2, C2, V = symbols('x y R w C R2 C2 V')

# Equations
eq1 = Eq(x*(I*w*C + 1/R + I*w*C2) + y*(-I*w*C2), I*w*C*V)
eq2 = Eq(x*(I*w*C2) + y*(-1/R2 - I*w*C2), 0)

# Solve for x, y
solution = solve((eq1, eq2), (x, y))

# Transfer function H = y / V
H = solution[y] / V

# Expected transfer function
H1 = R**2 / (R + 1/(I*w*C))**2

# Simplify 
H_simplified = H.simplify()
H1_simplified = H1.simplify()

# Substitute values: R=10^6, C=15*10^-12
H_subs = H_simplified.subs({
    R2: 4e6,
    C2: 5e-7,
})

pprint(H_subs)
pprint(H1_simplified)
```
